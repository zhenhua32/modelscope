```python
def split_chapter(text):
    """
    按章节(##)切分markdown文本
    
    Args:
        text (str): 输入的markdown文本
    
    Returns:
        list: 切分后的章节列表
    """
    import re
    
    # 使用正则表达式按 ## 开头的行进行切分
    # 保留分隔符，这样我们可以知道每个章节的标题
    chapters = re.split(r'^(## .+)$', text, flags=re.MULTILINE)
    
    # 过滤掉空字符串，并重新组合章节标题和内容
    result = []
    for i in range(1, len(chapters), 2):
        if i + 1 < len(chapters):
            chapter_title = chapters[i]
            chapter_content = chapters[i + 1]
            full_chapter = chapter_title + chapter_content
            result.append(full_chapter.strip())
        else:
            # 处理最后一个章节（可能没有后续内容）
            result.append(chapters[i].strip())
    
    return result


def split_page(text):
    """
    按页(###)切分章节文本，不包含##开头的第一段
    
    Args:
        text (str): 输入的章节文本
    
    Returns:
        list: 切分后的页面列表
    """
    import re
    
    # 先找到第一个 ## 行的结束位置
    lines = text.split('\n')
    start_index = 0
    
    # 跳过 ## 开头的第一段
    for i, line in enumerate(lines):
        if line.strip().startswith('## '):
            # 找到下一个非空行或者第一个 ### 行作为开始
            for j in range(i + 1, len(lines)):
                if lines[j].strip() and not lines[j].strip().startswith('## '):
                    start_index = j
                    break
            break
    
    # 从找到的位置开始重新组合文本
    remaining_text = '\n'.join(lines[start_index:])
    
    # 按 ### 切分
    pages = re.split(r'^(### .+)$', remaining_text, flags=re.MULTILINE)
    
    # 过滤掉空字符串，并重新组合页面标题和内容
    result = []
    for i in range(1, len(pages), 2):
        if i + 1 < len(pages):
            page_title = pages[i]
            page_content = pages[i + 1]
            full_page = page_title + page_content
            result.append(full_page.strip())
        else:
            # 处理最后一个页面（可能没有后续内容）
            result.append(pages[i].strip())
    
    # 如果没有找到任何 ### 标题，但有内容，则返回整个剩余内容
    if not result and remaining_text.strip():
        result.append(remaining_text.strip())
    
    return result
```

```python
# 测试数据
markdown_text = """
# 主标题

一些介绍内容

## 第一章
第一章的介绍内容
有很多内容

### 第一页
第一页的内容

### 第二页
第二页的内容

## 第二章
第二章的介绍内容

### 第一页
第二章第一页的内容
#### 有四级标题
和四级内容
##### 五级标题
和五级内容

### 第二页
第二章第二页的内容
"""

# 测试 split_chapter
chapters = split_chapter(markdown_text)
print("章节数量:", len(chapters))
for i, chapter in enumerate(chapters):
    print(f"第{i+1}章:\n{chapter}\n{'='*50}")

print("------")
# 测试 split_page
if chapters:
    pages = split_page(chapters[1])  # 切分第一章
    print("第一章页面数量:", len(pages))
    for i, page in enumerate(pages):
        print(f"第{i+1}页:\n{page}\n{'='*30}")
```

```python
# 这种方式看起来更优秀一些
import re

def split_chapter(text):
    """
    按章节(##)切分markdown文本
    
    Args:
        text (str): 输入的markdown文本
    
    Returns:
        list: 按章节切分后的文本数组
    """
    # 使用正则表达式按 ## 开头的行进行切分
    chapters = re.split(r'\n(?=##\s)', text.strip())
    
    # 过滤掉空字符串
    chapters = [chapter.strip() for chapter in chapters if chapter.strip()]
    
    return chapters
```
